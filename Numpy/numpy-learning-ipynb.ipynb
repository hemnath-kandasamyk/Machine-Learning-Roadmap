# 📊 3. MAE — Mean Absolute Error

Mean Absolute Error (MAE) is a **regression performance metric** used to measure how far the model's predictions are from the actual values.

### 📐 Formula

$$
MAE = \frac{1}{n}\sum_{i=1}^{n}|y_i-\hat{y}_i|
$$

### 🔍 Meaning of the symbols

| Symbol | Meaning |
|---|---|
| $n$ | Number of observations |
| $y_i$ | Actual value |
| $\hat{y}_i$ | Predicted value |
| $|y_i-\hat{y}_i|$ | Absolute error |
| $\sum$ | Sum of all errors |

### 🧠 In Simple Words

> **MAE = Average of the absolute differences between actual and predicted values.**

### 🔄 How MAE is calculated

```text
Actual & Predicted
        ↓
Calculate Error
        ↓
Take Absolute Value
        ↓
Calculate Average
        ↓
       MAE

````python
mae = np.mean(np.abs(actual - predicted))

print("MAE:", mae)
````

# 📊 4. MSE — Mean Squared Error

Mean Squared Error (MSE) is a **regression performance metric** used to measure the average squared difference between actual and predicted values.

### 📐 Formula

$$
MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

### 🔍 Meaning of the Symbols

| Symbol | Meaning |
|---|---|
| $n$ | Number of observations |
| $y_i$ | Actual value |
| $\hat{y}_i$ | Predicted value |
| $y_i-\hat{y}_i$ | Prediction error |
| $(y_i-\hat{y}_i)^2$ | Squared error |
| $\sum$ | Sum of all squared errors |

### 🧠 In Simple Words

> **MSE = Average of the squared differences between actual and predicted values.**

### 🔄 How MSE is Calculated

```text
Actual & Predicted
        ↓
Calculate Error
        ↓
Square the Error
        ↓
Calculate Average
        ↓
       MSE

````python
mse = np.mean((actual - predicted) ** 2)

print("MSE:", mse)
````

# 📊 5. RMSE — Root Mean Squared Error

Root Mean Squared Error (RMSE) is a **regression performance metric** that measures the typical difference between actual and predicted values.

RMSE is obtained by taking the **square root of MSE**.

### 📐 Formula

$$
RMSE = \sqrt{\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2}
$$

Another way to remember it:

$$
\boxed{RMSE = \sqrt{MSE}}
$$

### 🔍 Meaning of the Symbols

| Symbol | Meaning |
|---|---|
| $n$ | Number of observations |
| $y_i$ | Actual value |
| $\hat{y}_i$ | Predicted value |
| $y_i-\hat{y}_i$ | Prediction error |
| $(y_i-\hat{y}_i)^2$ | Squared error |
| $\sum$ | Sum of squared errors |
| $\sqrt{}$ | Square root |

### 🧠 In Simple Words

> **RMSE = Square root of the average squared prediction errors.**

### 🔄 How RMSE is Calculated

```text
Actual & Predicted
        ↓
Calculate Error
        ↓
Square the Error
        ↓
Calculate Mean
        ↓
       MSE
        ↓
Take Square Root
        ↓
      RMSE

````python
rmse = np.sqrt(np.mean((actual - predicted) ** 2))

print("RMSE:", rmse)
````

## 📊 MAE vs MSE vs RMSE

| Metric | Formula | Same Unit as Target? | Large Error Sensitivity |
|---|---|---|---|
| **MAE** | `mean(abs(error))` | ✅ Yes | Low |
| **MSE** | `mean(error²)` | ❌ No | High |
| **RMSE** | `sqrt(MSE)` | ✅ Yes | High |

### Quick Understanding

```text
MAE
 ↓
Average absolute error

MSE
 ↓
Average squared error
 ↓
Large errors get heavily penalized

RMSE
 ↓
Square root of MSE
 ↓
Returns to original unit

# 📊 6. R² Score — R-Squared

R² (R-Squared), also called the **Coefficient of Determination**, is a regression performance metric.

It tells us **how well the model explains the variation in the target variable**.

### 📐 Formula

$$
R^2 = 1 - \frac{SS_{res}}{SS_{tot}}
$$

Where:

$$
SS_{res} = \sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

$$
SS_{tot} = \sum_{i=1}^{n}(y_i-\bar{y})^2
$$

Therefore:

$$
\boxed{
R^2 =
1 -
\frac{
\sum(y_i-\hat{y}_i)^2
}{
\sum(y_i-\bar{y})^2
}
}
$$

### 🔍 Meaning of the Symbols

| Symbol | Meaning |
|---|---|
| $y_i$ | Actual value |
| $\hat{y}_i$ | Predicted value |
| $\bar{y}$ | Mean of actual values |
| $SS_{res}$ | Residual Sum of Squares |
| $SS_{tot}$ | Total Sum of Squares |
| $n$ | Number of observations |

### 🧠 In Simple Words

> **R² tells us how much of the variation in the target variable is explained by the model.**

For example:

```text
R² = 0.80

````python
import numpy as np

actual = np.array([100, 200, 300, 400])
predicted = np.array([90, 210, 280, 390])

# Mean of actual values
mean_actual = np.mean(actual)

# Residual Sum of Squares
ss_res = np.sum((actual - predicted) ** 2)

# Total Sum of Squares
ss_tot = np.sum((actual - mean_actual) ** 2)

# R² Score
r2 = 1 - (ss_res / ss_tot)

print("Actual Values:    ", actual)
print("Predicted Values: ", predicted)
print("Mean Actual:      ", mean_actual)
print("SS_res:           ", ss_res)
print("SS_tot:           ", ss_tot)
print("R² Score:         ", r2)
````

## 🎯 Interpretation

| R² Score | General Interpretation |
|---:|---|
| `1.0` | Perfect prediction |
| `0.9` | Very strong fit |
| `0.8` | Strong fit |
| `0.5` | Moderate fit |
| `0.2` | Weak fit |
| `0.0` | No improvement over predicting the mean |
| `< 0` | Worse than the mean baseline |

> ⚠️ These are general guidelines, not universal thresholds. What counts as a good R² depends on the dataset and application.

### Important

R² can be **negative**.

A negative R² means the model performs worse than simply predicting the mean of the actual target values.

# 📊 Classification Performance Metrics

Classification metrics are used to evaluate how well a classification model predicts **classes or categories**.

Common classification metrics:

- Accuracy
- Precision
- Recall
- F1 Score

### 🧩 Confusion Matrix

Before calculating these metrics, understand four terms:

| Term | Meaning |
|---|---|
| **TP** | True Positive — correctly predicted positive |
| **TN** | True Negative — correctly predicted negative |
| **FP** | False Positive — predicted positive, actually negative |
| **FN** | False Negative — predicted negative, actually positive |

```text
                    Predicted
                  Positive  Negative
Actual Positive     TP        FN
       Negative     FP        TN

````python
import numpy as np

# Actual class labels
actual = np.array([1, 1, 1, 0, 0, 0, 1, 0])

# Model predictions
predicted = np.array([1, 1, 0, 0, 1, 0, 1, 0])

# Confusion Matrix values
tp = np.sum((actual == 1) & (predicted == 1))
tn = np.sum((actual == 0) & (predicted == 0))
fp = np.sum((actual == 0) & (predicted == 1))
fn = np.sum((actual == 1) & (predicted == 0))

print("True Positive :", tp)
print("True Negative :", tn)
print("False Positive:", fp)
print("False Negative:", fn)
````

# 1️⃣ Accuracy

Accuracy measures the percentage of predictions that the model classified correctly.

### 📐 Formula

$$
Accuracy = \frac{TP + TN}{TP + TN + FP + FN}
$$

### 🧠 In Simple Words

> **Accuracy = Number of correct predictions / Total predictions**

### 🎯 Interpretation

- `1.0` → 100% correct
- `0.0` → 0% correct
- Higher Accuracy → Generally better

### ⚠️ Important

Accuracy can be misleading when the classes are highly imbalanced.

For example:

```text
100 patients
95 → Negative
5  → Positive

````python
accuracy = (tp + tn) / (tp + tn + fp + fn)

print("Accuracy:", accuracy)
print("Accuracy (%):", accuracy * 100)
````

# 2️⃣ Precision

Precision tells us how many of the samples predicted as **Positive** were actually Positive.

### 📐 Formula

$$
Precision = \frac{TP}{TP + FP}
$$

### 🧠 In Simple Words

> **Of everything the model predicted as Positive, how much was actually Positive?**

### Example

If:

```text
TP = 3
FP = 1

````python
precision = tp / (tp + fp)

print("Precision:", precision)
````

# 3️⃣ Recall

Recall measures how many of the **actual Positive samples** were correctly identified by the model.

### 📐 Formula

$$
Recall = \frac{TP}{TP + FN}
$$

### 🧠 In Simple Words

> **Out of all actual Positive cases, how many did the model successfully find?**

### Example

If:

```text
TP = 3
FN = 1

````python
recall = tp / (tp + fn)

print("Recall:", recall)
````

# 4️⃣ F1 Score

F1 Score combines **Precision and Recall** into a single metric.

### 📐 Formula

$$
F1 = 2 \times
\frac{Precision \times Recall}
{Precision + Recall}
$$

### 🧠 In Simple Words

> **F1 Score is the harmonic mean of Precision and Recall.**

It is useful when we want a balance between:

```text
Precision
    +
Recall
    ↓
F1 Score

````python
f1 = 2 * (precision * recall) / (precision + recall)

print("F1 Score:", f1)
````

# 🧠 Classification Metrics — Quick Recall

| Metric | Formula | Main Question |
|---|---|---|
| **Accuracy** | `(TP + TN) / Total` | How many predictions are correct? |
| **Precision** | `TP / (TP + FP)` | When model says Positive, how often is it right? |
| **Recall** | `TP / (TP + FN)` | How many actual Positives did it find? |
| **F1 Score** | `2PR / (P + R)` | How well are Precision and Recall balanced? |

### 🔥 Remember

```text
Accuracy  → Overall correctness
Precision → Positive predictions correctness
Recall    → Positive cases found
F1 Score  → Precision + Recall balance
